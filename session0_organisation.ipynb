# Organisation de RCP209


<dl style='margin: 20px 0;'>
<dt>L’unité d’enseignement RCP09 est planifiée sur 15 séances (2h de cours + 2h de TP):</dt>
<dd>
- En cours du soir au CNAM (semestre 1): les séances ont lieu au CNAM Paris de 17h30 à 21h30, à raison de une par semaine. L’emploi du temps se trouve sur [ce lien](http://emploi-du-temps.cnam.fr/)  
- En FOAD (semestre 2): il convient de suivre les contenus à son rythme. Nous recommendons toutefois un travail régulier suivant le tempo d’une séance (cours+tp) par semaine. Des séances de regroupement (points cours et/ou échanges en visioconférence) sont organisées périodiquement sur la plateforme ENF du CNAM.  


</dd>

</dl>

**Important** Si vous suivez l’UE RCP209 du CNAM dans le cadre d’une formation financée par un organisme (employeur, France Travail, Conseil régional, etc.), il est indispensable de passer par [http://lecnam.net](http://lecnam.net) pour accéder à ces supports (plutôt que de consulter directement ce site public) afin que des indicateurs de suivi puissent être transmis à cet organisme.

## Evaluation


<dl style='margin: 20px 0;'>
<dt>La note finale est la moyenne (non pondérée) de deux notes:</dt>
<dd>
- la note d’examen final: l’examen est un examen sur table. Pendant l’examen vous pouvez consulter comme **seul document écrit** deux feuilles A4 (recto-verso) **écrites à la main** ; l’objectif est de vous inciter à faire vous-même la synthèse que vous utilisez pendant l’examen. La planification des examens est accessible [ici](http://www.cnam-paris.fr/suivre-ma-scolarite/) (rubrique « Examens »).  
- la note au projet de l’UE: vous trouverez [ici les informations concernant les projets](session0_projets.ipynb#chap-projets). Vous êtes encouragés à **proposer vous-même** un sujet de projet qui vous intéresse particulièrement.  


</dd>

</dl>
