# RCP209 - Apprentissage statistique : modélisation décisionnelle et apprentissage profond

Ce site regroupe les supports écrits de l’unité d’enseignement [RCP209](https://formation.cnam.fr/rechercher-par-discipline/apprentissage-statistique-modelisation-decisionnelle-et-apprentissage-profond-208748.kjsp),
proposé par le [département d’informatique du Conservatoire National des Arts et Métiers (CNAM)](https://informatique.cnam.fr/fr/).
Cette unité d’enseignement présente les modèles prédictifs pour l’analyse de données, notamment les arbres de décisions, les machines à vecteurs de support et les réseaux de neurones profonds (deep learning).

L’ensemble des contenus est libre et directement accessible via la barre de navigation.
