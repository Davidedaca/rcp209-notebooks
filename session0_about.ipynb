# A propos de RCP209

RCP209 est une unité d’enseignement (UE) du Conservatoire national des arts et métiers (CNAM) visant à transmettre les cadres méthodologiques (et outils logiciels associés) pour comprendre et mettre en oeuvre des modèles d’IA modernes.

Cette UE est ouverte au CNAM Paris en cours du soir au premier semestre de l’année universitaire et en formation à distance (FOD ou FOAD) nationale au second semestre.
L’accès aux informations administratives et à l’inscription se trouvent sur [la fiche descriptive de l’UE](https://formation.cnam.fr/rechercher-par-discipline/apprentissage-statistique-modelisation-decisionnelle-et-apprentissage-profond-208748.kjsp) (onglet « Informations pratiques »), ainsi que sur [la page BéDéO](https://bedeo.cnam.fr/public/unite/view/RCP209/2025-01-05).

Le contenu pédagogique est laissé en libre accès sur ce site.
Le reste (séances de cours, accès aux ressources auditeurs CNAM comme la plateforme ENF, inscription à l’examen pour validation des ECTS, etc.) requiert une inscription à l’UE au CNAM.

## Pré-requis

Cette UE s’adresse à un public ayant des connaissance en informatique (programmation, bases de données) et en mathématiques appliquées (niveau bac+5).
Il est conseillé d’avoir suivi les UE NFP106 et RCP208 au préalable.
